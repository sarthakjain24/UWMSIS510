```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('movieLens.txt', sep='|', header=None, quotechar="\"")
```

```python
df.head()
```

```python
df.columns  = ['ID', 'Title', 'ReleaseDate','VideoReleaseDate', 'IMDB', 'Unknown', 'Action',
               'Adventure','Animation','Childrens','Comedy', 'Crime', 'Documentary', 'Drama',
               'Fantasy', 'FilmNoir','Horror', 'Musical', 'Mystery', 'Romance', 'SciFi', 'Thriller',
               'War', 'Western']
```

```python
df.head()
```

```python
df.shape
```

```python
df.drop_duplicates(subset=['Title'], inplace=True, keep='first')
```

```python
df.shape
```

```python
X = df.drop(['ID','Title','ReleaseDate','VideoReleaseDate','IMDB','Unknown'], axis = 1)
X.head()
```

```python
kmeans = KMeans(n_clusters=10, random_state=42)
kmeans.fit(X)
```

```python
kmeans.labels_
```

```python
X['cluster_ids'] = kmeans.labels_
X.head()
```

```python
X[X['cluster_ids'] == 1].mean()
```

```python
df['cluster_ids'] = kmeans.labels_ # adding a new column in df called cluster_ids
df[df['cluster_ids'] == 4]
```
