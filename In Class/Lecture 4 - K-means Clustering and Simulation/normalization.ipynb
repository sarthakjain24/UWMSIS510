```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('WHO (4).csv')
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X = df.drop(['Country','Region'], axis=1)
X.head()
```

```python
X_std = (X - X.mean()) / X.std()
X_std.head()
```


kmeans = KMeans(n_clusters=6, random_state=42)
kmeans.fit(X_std)

```python
kmeans = KMeans(n_clusters=6, random_state=42)
kmeans.fit(X_std)
```

```python
df['cluster_ids'] = kmeans.labels_
```
