```python
import pandas as pd
from sklearn.model_selection import train_test_split
import statsmodels.formula.api as smf
```

```python
df = pd.read_csv("Framingham-2-1.csv")
```

```python
df.describe()
```

```python
df[df['cigsPerDay'] == 70]
```

```python
#missing values, dropping any row that has no values
df = df.dropna()
```

```python
df['TenYearCHD'].value_counts()
```

```python
df_train, df_test = train_test_split(df, test_size=0.2, random_state=42,stratify=df['TenYearCHD'])
```

```python
logistic_regression_1 = smf.logit(formula='TenYearCHD ~ age + totChol + sysBP + male', data=df_train).fit()
logistic_regression_1.summary()
```

```python
independent_vars = list(df_train.columns)
independent_vars.remove('TenYearCHD')
```

```python
" + ".join(independent_vars)
```

```python
logitformula = "TenYearCHD ~ " + " + ".join(independent_vars)
```

```python
logistic_regression_final = smf.logit(formula=logitformula, data=df_train).fit()
print(logistic_regression_final.summary())
```

```python
logistic_regression_final.predict(df_test)
```
