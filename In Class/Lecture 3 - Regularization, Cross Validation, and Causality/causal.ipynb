```python
import numpy as np
import statsmodels.formula.api as smf
import pandas as pd
from linearmodels.iv import IV2SLS
```

```python
df = pd.read_csv('smartphone_sales-1-1-1.csv')

model = smf.ols('sales ~ Prices + display_size + primary_camera + storage + battery', data=df).fit()
print(model.summary())
```

```python
iv_regression = IV2SLS.from_formula("sales ~ 1+ display_size +	primary_camera + storage +	battery +	[Prices~CNYExchangeRateIndex]", data =df)
iv_regression_results = iv_regression.fit()
print(iv_regression_results)
```
